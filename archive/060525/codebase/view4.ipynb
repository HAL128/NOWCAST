```python
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib

# データの読み込み
df = pd.read_csv('../data/G1_yoy.csv')

# 日付をdatetime型に変換
df['DATE'] = pd.to_datetime(df['DATE'])

# グラフの設定
plt.figure(figsize=(12, 6))
plt.plot(df['DATE'], df['MEAN_YOY'], label='MEAN_YOY')
plt.plot(df['DATE'], df['SUM_YOY'], label='SUM_YOY')

# グラフの装飾
plt.title('MEAN_YOYとSUM_YOYの推移')
plt.xlabel('日付')
plt.ylabel('前年比')
plt.grid(True)
plt.legend()

# x軸の日付を45度回転
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
