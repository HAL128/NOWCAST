```python
import pandas as pd
import re
```

```python
# データフレームの読み込み例（ファイル名や区切り文字は適宜修正）
df = pd.read_csv('../data/jp_data/gaku-mk2511.csv')
```

```python
# 年を保持する変数
current_year = None

def fill_year(date):
    global current_year
    # 年が含まれている場合（YYYY-...）
    if isinstance(date, str) and len(date) >= 10 and date[:4].isdigit():
        current_year = date[:4]
        return date
    # 年が抜けている場合（-MM-DD）
    elif isinstance(date, str) and date.startswith('-'):
        return f"{current_year}{date}"
    else:
        return date

# date列に適用
df['date'] = df['date'].apply(fill_year)

# 結果確認
df
```
